### first we used tavily search tool. Now we will write out own custom tool and will use it also along with builtin tavily tool.

```python
@tool
def get_weather_data(city: str) -> str:
    """
    Fetch current weather information for a city.
    """

    url = (
        f"https://api.weatherstack.com/current?"
        f"access_key={WEATHERSTACK_API_KEY}&query={city}"
    )

    response = requests.get(url)

    data = response.json()

    if "current" not in data:
        return f"Could not fetch weather data for {city}"

    return (
        f"City: {city}\n"
        f"Temperature: {data['current']['temperature']}°C\n"
        f"Weather: {data['current']['weather_descriptions'][0]}\n"
        f"Humidity: {data['current']['humidity']}%"
    )
```

```python

# ==========================================
# LLM - Initialize LLM
# ==========================================

llm = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash",
    google_api_key=GEMINI_API_KEY
)
```

```python
# ==========================================
# TOOLS
# ==========================================

tools = [search_tool, get_weather_data]
```

```python
system_prompt = """
You are a helpful AI assistant.

You have access to:
1. A web search tool for current information.
2. A weather tool for current weather information.

Use the weather tool when the user asks about weather.
Use the search tool when current information is needed.

Answer the user clearly and concisely.
"""
```

```python
# ==========================================
# CREATE AGENT
# ==========================================

agent = create_agent(
    model=llm,
    tools=tools,
    system_prompt=system_prompt
)
```

```python
# ==========================================
# EXECUTOR
# ==========================================

response = agent.invoke({
    "messages": [
        {
            "role": "user",
            "content": "Find the capital of Pakistan"
                       "and then find its current weather."
        }
    ]
})
```

```python
print(response["messages"][-1].content)
```
